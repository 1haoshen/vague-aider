# MVISU-Bench 数据集统计与分类

这个 Notebook 用于统计 MVISU-Bench 数据集中的任务类型并对指令进行分类。


```python
import json
import os
import pandas as pd
from IPython.display import display

pd.set_option('display.max_colwidth', None)
```

```python
data_dir = '.'
files = [f for f in os.listdir(data_dir) if f.endswith('_mvisu_bench.json')]
```

```python
tasks_list = []

for file in files:
    with open(os.path.join(data_dir, file), 'r') as f:
        data = json.load(f)
        language = file.split('_')[0]  # cn or en
        for task in data:
            tasks_list.append({
                'language': language,
                'ID': task['ID'],
                'instruction': task['Instruction'],
                'TaskType': task['TaskType'],
                'APP': ', '.join(task.get('APP', [])),
                'APPType': ', '.join(task.get('APPType', []))
            })

df = pd.DataFrame(tasks_list)
```

```python
print("任务类型统计:")
print(df['TaskType'].value_counts())

print("\n语言统计:")
print(df['language'].value_counts())

print(f"\n总任务数: {len(df)}")
```

## 说明

- 使用 pandas 进行数据处理。
- 任务类型统计显示每个类型的任务数量。
- 添加了查询和分组计数的示例单元格。
- 新增显示所有任务列表的单元格，包括具体指令。
- 已设置 pandas 显示选项，使 instruction 列完整显示而不省略。
- 已添加样式，使 instruction 列左对齐显示。
- 您可以运行以上单元格来查看结果，或修改查询参数进行自定义分析。


```python
# 示例查询：过滤特定任务类型的任务，并显示指令（instruction 作为第三列） 
specific_tasktype = 'Multi-App'
filtered = df.query("TaskType == @specific_tasktype")
print(f"{specific_tasktype} 中的任务数: {len(filtered)}")
styled = filtered.style.set_properties(subset=['instruction'], **{'text-align': 'left'})
display(styled)
#    python run.py --instruction "转发一张高达照片给抖音好友sey_yra。"
```

```python
# 示例计数：按类型和语言分组计数
grouped = df.groupby(['TaskType', 'language']).size().reset_index(name='count')
print(grouped)
```

```python
# 显示所有任务的列表，包括具体指令（instruction 作为第三列）
styled = df.style.set_properties(subset=['instruction'], **{'text-align': 'left'})
display(styled)
```

```python
unique_apps = set()
for apps_str in df['APP']:
    if apps_str:
        apps = apps_str.split(', ')
        unique_apps.update(apps)
print(f"所有指令涉及的独特app总数: {len(unique_apps)}")
```
