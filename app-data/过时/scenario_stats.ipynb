# 指令數據集場景統計與分類

這個 Notebook 用於統計 Mobile-Eval-E 數據集中的場景並對任務進行分類。


```python
import json
import os
import pandas as pd
from IPython.display import display

pd.set_option('display.max_colwidth', None)
```

```python
data_dir = '.'
files = [f for f in os.listdir(data_dir) if f.endswith('.json')]
```

```python
tasks_list = []

for file in files:
    with open(os.path.join(data_dir, file), 'r') as f:
        data = json.load(f)
        scenario = data['scenario']
        for task in data['tasks']:
            tasks_list.append({
                'scenario': scenario,
                'task_id': task['task_id'],
                'instruction': task['instruction'],
                'type': task['type'],
                'apps': ', '.join(task['apps'])
            })

df = pd.DataFrame(tasks_list)
```

```python
print("場景統計:")
print(df['scenario'].value_counts())

print("\n任務類型分類:")
print(df['type'].value_counts())

print(f"\n總任務數: {len(df)}")
```

## 說明

- 現在使用 pandas 進行數據處理。
- 場景統計顯示每個場景下的任務數量。
- 任務類型分類統計 single_app 和 multi_app 等類型。
- 添加了查詢和分組計數的示例單元格。
- 新增顯示所有任務列表的單元格，包括具體指令。
- 已設置 pandas 顯示選項，使 instruction 列完整顯示而不省略。
- 已添加樣式，使 instruction 列左對齊顯示。
- 您可以運行以上單元格來查看結果，或修改查詢參數進行自定義分析。


```python
# 示例查詢：過濾特定場景的任務，並顯示指令（instruction 作為第三列）
specific_scenario = 'RESTAURANT_RECOMMENDATION'
filtered = df.query("scenario == @specific_scenario")
print(f"{specific_scenario} 中的任務數: {len(filtered)}")
styled = filtered.style.set_properties(subset=['instruction'], **{'text-align': 'left'})
display(styled)
```

```python
# 示例計數：按類型和場景分組計數
grouped = df.groupby(['type', 'scenario']).size().reset_index(name='count')
print(grouped)
```

```python
# 顯示所有任務的列表，包括具體指令（instruction 作為第三列）
styled = df.style.set_properties(subset=['instruction'], **{'text-align': 'left'})
display(styled)

```
