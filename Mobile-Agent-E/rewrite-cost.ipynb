```python
import json
import pandas as pd
from IPython.display import display
```

```python
# Process LLM-output.json for per-instruction rewrite cost averages

# Load LLM-output.json
with open('LLM-output.json', 'r', encoding='utf-8') as f:
    llm_data = json.load(f)

# Convert to DataFrame
llm_df = pd.DataFrame(llm_data)

# Extract from 'token_usage'
if 'token_usage' in llm_df.columns:
    llm_df['prompt_tokens'] = llm_df['token_usage'].apply(lambda x: x.get('prompt_tokens', 0) if isinstance(x, dict) else 0)
    llm_df['completion_tokens'] = llm_df['token_usage'].apply(lambda x: x.get('completion_tokens', 0) if isinstance(x, dict) else 0)
    llm_df['total_tokens'] = llm_df['token_usage'].apply(lambda x: x.get('total_tokens', 0) if isinstance(x, dict) else 0)
else:
    print("No 'token_usage' column found in LLM-output.json.")

# Calculate per-instruction averages (ignoring zeros)
def avg_ignore_zero(series):
    non_zero = series[series > 0]
    return non_zero.mean() if not non_zero.empty else 0

llm_prompt_avg = avg_ignore_zero(llm_df['prompt_tokens'])
llm_completion_avg = avg_ignore_zero(llm_df['completion_tokens'])
llm_total_avg = avg_ignore_zero(llm_df['total_tokens'])
llm_duration_avg = avg_ignore_zero(llm_df.get('duration_seconds', pd.Series()))

# Create DataFrame for display
llm_stats_df = pd.DataFrame({
    'Metric': ['avg_prompt_tokens_per_instruction', 'avg_completion_tokens_per_instruction', 'avg_total_tokens_per_instruction', 'avg_duration_seconds_per_instruction'],
    'Value': [llm_prompt_avg, llm_completion_avg, llm_total_avg, llm_duration_avg]
})

# Debug prints
print("Columns in LLM-output:", llm_df.columns.tolist())
if not llm_df.empty:
    print("Sample token_usage:", llm_df['token_usage'].iloc[0])

# Display the table
print("Per-Instruction Rewrite Cost Averages from LLM-output.json:")
display(llm_stats_df)
```
